## 1.Loading and Preprocessing

```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.preprocessing import StandardScaler

data=load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df["Target"] = data.target

df.head()
```

```python
x = df.drop("Target", axis=1)
y = df["Target"]
```

## 2.Classification Algorithm Implementation

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
#Logistic Regression
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)

model.fit(x_train, y_train)
x_new = pd.DataFrame([[
    17.99,10.38,122.80,1001.0,    
    0.11840,   
    0.27760,  
    0.3001,    
    0.14710,   
    0.2419,    
    0.07871,   

    1.095,    
    0.9053,    
    8.589,     
    153.40,    
    0.006399,  
    0.04904,   
    0.05373,   
    0.01587,   
    0.03003,   
    0.006193,  

    25.38,     
    17.33,    
    184.60,    
    2019.0,    
    0.1622,    
    0.6656,    
    0.7119,    
    0.2654,    
    0.4601,    
    0.11890    
]],
    columns=df.drop("Target",axis=1).columns
     )
x_new = scaler.transform(x_new)

probability = model.predict_proba(x_new)

logistic_prediction= model.predict(x_new)


print("Probability:", probability)
print("Prediction:", logistic_prediction)
print("Prediction:", data.target_names[logistic_prediction])
```

### Logistic Regression predicts the probability that a sample belongs to a particular class

```python
#Decision Tree
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state=42)

model.fit(x_train, y_train)

decision_tree_pred = model.predict(x_new)

print("Prediction:", data.target_names[decision_tree_pred])
```

### Decision Tree makes prediction for a new data by asking a series of YES or NO questions.

```python
#Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state=42)

model.fit(x_train, y_train)

random_forest_pred  = model.predict(x_new)

print("Prediction:", data.target_names[random_forest_pred])
```

### Random Forest combines multiple Decision Trees and uses their combined predictions to classify the data.

```python
#Support Vector Machine (SVM)
from sklearn.svm import SVC

model = SVC()

model.fit(x_train, y_train)

svm_pred = model.predict(x_new)

print("Prediction:", data.target_names[svm_pred])
```

### SVM finds a boundary that separates different classes as clearly as possible.

```python
#KNN
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=5)

model.fit(x_train, y_train)

knn_pred = model.predict(x_new)

print("Prediction:", data.target_names[knn_pred])
```

### k-NN classifies a new sample based on the classes of its nearest data points. It is suitable because the breast cancer dataset contains numerical features that can be compared based on their distances.

## 3.Model Comparison

```python
from sklearn.metrics import accuracy_score
```

```python
# Logistic Regression
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
logistic_test_pred = model.predict(x_test)

# Decision Tree
model = DecisionTreeClassifier(random_state=42)
model.fit(x_train, y_train)
decision_tree_test_pred = model.predict(x_test)

# Random Forest
model = RandomForestClassifier(random_state=42)
model.fit(x_train, y_train)
random_forest_test_pred = model.predict(x_test)

# SVM
model = SVC()
model.fit(x_train, y_train)
svm_test_pred = model.predict(x_test)

# KNN
model = KNeighborsClassifier(n_neighbors=5)
model.fit(x_train, y_train)
knn_test_pred = model.predict(x_test)

# Accuracy
print("Logistic Regression:", accuracy_score(y_test, logistic_test_pred))
print("Decision Tree:", accuracy_score(y_test, decision_tree_test_pred))
print("Random Forest:", accuracy_score(y_test, random_forest_test_pred))
print("SVM:", accuracy_score(y_test, svm_test_pred))
print("KNN:", accuracy_score(y_test, knn_test_pred))
```

### SVM performed the best with an accuracy of 98.25%.
### Decision Tree and KNN performed the worst, both with 94.74% accuracy.
