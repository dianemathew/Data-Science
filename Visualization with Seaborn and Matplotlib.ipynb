## 1. Countplot: Passenger Count by Embarked Port
### ● Create a countplot to display the number of passengers who boarded from each port (embarked), separated by sex.

```python
sns.countplot(data=df, x="Embarked", hue="Sex")

plt.xlabel("Embarked Port")
plt.ylabel("Passenger Count")
plt.title("Passenger Count by Embarked Port and Sex")
plt.show()
```

#### Observation: Most passengers boarded from Southampton (S), followed by Cherbourg (C) and Queenstown (Q). The number of male passengers is higher than female passengers across the ports.

## 2. Swarmplot: Age Distribution across Classes
### ● Create a swarmplot to visualize the distribution of age across passenger classes (pclass), categorized by sex.

```python
sns.swarmplot(data=df, x="Pclass", y="Age", hue="Sex")

plt.xlabel("Passenger Class")
plt.ylabel("Age")
plt.title("Age Distribution across Passenger Classes by Sex")
plt.show()
```

#### Observation: Passengers in 1st class generally have a wider age range, while 3rd class contains many younger passengers. Both male and female passengers are present across all classes.

## 3. Pie Chart: Gender Distribution
### ● Create a pie chart using Matplotlib to show the percentage of male vs. female passengers. - Add percentage labels using autopct - Use explode for one slice to highlight it.

```python
gender_count = df["Sex"].value_counts()

plt.pie(gender_count.values, labels=gender_count.index,autopct="%1.1f%%",explode=[0.1,0])
plt.title("Gender Distribution")
```

#### Observation: The dataset contains more male passengers than female passengers, showing a higher proportion of males.

## 4. Heatmap: Correlation Matrix
### ● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,and survived.

```python
corr = df[["Age", "Fare", "Pclass", "SibSp", "Parch", "Survived"]].corr()

sns.heatmap(corr, annot=True)

plt.title("Correlation Matrix")
plt.show()
```

#### Observation: Pclass and Fare show a noticeable negative correlation.

## 5. Violin Plot: Fare by Class and Gender
### ● Create a violin plot to visualize fare distribution across classes, separated by gender.

```python
sns.violinplot(data=df, x="Pclass", y="Fare", hue="Sex")

plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Fare Distribution by Class and Gender")
plt.show()
```

#### Observation: 1st-class passengers generally paid higher fares than 2nd- and 3rd-class passengers.

## 6. Scatter Plot: Age vs Fare
### ● Create a scatter plot to explore the relationship between age and fare, colored by survival status.
### ● Use a seaborn scatterplot with hue='survived'. Each plot must have titles, labels, and legends where needed.

```python
sns.scatterplot(data=df, x="Age", y="Fare", hue="Survived")

plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare by Survival Status")
plt.legend(title="Survived")
plt.show()
```

#### Observation: There is no strong clear relationship between age and fare.
